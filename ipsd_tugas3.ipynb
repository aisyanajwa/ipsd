Secara umum, pendidikan orang tua tidak secara signifikan membantu siswa untuk berprestasi baik dalam ujian. Terdapat peningkatan pada median skor rata-rata siswa dari tingkat pendidikan orang tua some college hingga master's degree, meskipun perubahannya tergolong tipis, menyoroti adanya peningkatan pada median skor rata-rata berdasarkan tingkat pendidikan orang tua, dan mengakui bahwa meskipun perubahan tersebut ada, dampaknya tidak terlalu besar.

```python
Group_data2 = df.groupby('race/ethnicity')
# rata rata dicari
mean_math = Group_data2['math score'].mean()
mean_reading = Group_data2['reading score'].mean()
mean_writing = Group_data2['writing score'].mean()

plt.figure(figsize=(20, 8))

# Math scores
plt.subplot(1, 3, 1)
sns.barplot(x=mean_math.index, y=mean_math.values, palette='husl')
plt.title('Rata-rata Skor Matematika', size=20)
plt.xlabel('Ras/Etnisitas')
plt.ylabel('Skor Rata-rata')
for i in range(len(mean_math)):
    plt.text(i, mean_math.values[i], round(mean_math.values[i], 2), ha='center', va='bottom')

# Reading scores
plt.subplot(1, 3, 2)
sns.barplot(x=mean_reading.index, y=mean_reading.values, palette='husl')
plt.title('Rata-rata Skor Membaca', size=20)
plt.xlabel('Ras/Etnisitas')
plt.ylabel('Skor Rata-rata')
for i in range(len(mean_reading)):
    plt.text(i, mean_reading.values[i], round(mean_reading.values[i], 2), ha='center', va='bottom')

# Writing scores
plt.subplot(1, 3, 3)
sns.barplot(x=mean_writing.index, y=mean_writing.values, palette='husl')
plt.title('Rata-rata Skor Menulis', size=20)
plt.xlabel('Ras/Etnisitas')
plt.ylabel('Skor Rata-rata')
for i in range(len(mean_writing)):
    plt.text(i, mean_writing.values[i], round(mean_writing.values[i], 2), ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

Siswa dari Kelompok A telah meraih nilai terendah.

### 3. Lakukan uji korelasi:

```python
# korelasi
correlation_matrix = df[['math score', 'reading score', 'writing score']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='OrRd', fmt='.2f')
plt.title('Korelasi Skor')
plt.show()
```

Korelasi antar skor menunjukkan angka yang tinggi, siswa yang memiliki skor tinggi dalam satu subjek cenderung juga memiliki skor tinggi di subjek lainnya.
